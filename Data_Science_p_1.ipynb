```python
# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 2: Load Dataset
df = pd.read_csv("/content/Diwali Sales Data.csv", encoding='unicode_escape')
df
```

```python
# Step 3: Basic Info
print("Shape:", df.shape)
print("\nHead:\n", df.head())
print("\nInfo:\n",df.info())
```

```python
# Step 4: Data Cleaning
# Drop unwanted columns (ignore error if not present)
df.drop(['Status', 'unnamed1'], axis=1, inplace=True, errors='ignore')
```

```python
# Drop null values
df.dropna(inplace=True)
df
```

```python
# Convert Amount to integer safely
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')
df.dropna(inplace=True)
df['Amount'] = df['Amount'].astype(int)
df
```

```python
print("\nAfter Cleaning:\n")
print(df.info())
```

```python
# 📊 EXPLORATORY DATA ANALYSIS (EDA)
```

```python
sns.set(style="whitegrid")
```

```python
# 1. Gender Analysis
gender_sales = df.groupby('Gender')['Amount'].sum().sort_values(ascending=False)

plt.figure()
sns.barplot(x=gender_sales.index, y=gender_sales.values)
plt.title("Total Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Amount")
plt.show()
```

```python
# 2. Age Group Analysis
age_sales = df.groupby('Age Group')['Amount'].sum().sort_values(ascending=False)

plt.figure()
sns.barplot(x=age_sales.index, y=age_sales.values)
plt.title("Sales by Age Group")
plt.xticks(rotation=45)
plt.show()
```

```python
# 3. Marital Status Analysis
marital_sales = df.groupby('Marital_Status')['Amount'].sum()

plt.figure()
sns.barplot(x=marital_sales.index, y=marital_sales.values)
plt.title("Sales by Marital Status")
plt.show()
```

```python
# 4. State Analysis (Top 10)
state_sales = df.groupby('State')['Amount'].sum().sort_values(ascending=False).head(10)

plt.figure()
sns.barplot(x=state_sales.values, y=state_sales.index)
plt.title("Top 10 States by Sales")
plt.show()
```

```python
# 5. Occupation Analysis
occ_sales = df.groupby('Occupation')['Amount'].sum().sort_values(ascending=False)

plt.figure()
sns.barplot(x=occ_sales.values, y=occ_sales.index)
plt.title("Sales by Occupation")
plt.show()
```

```python
# 6. Product Category Analysis
prod_sales = df.groupby('Product_Category')['Amount'].sum().sort_values(ascending=False)

plt.figure()
sns.barplot(x=prod_sales.values, y=prod_sales.index)
plt.title("Sales by Product Category")
plt.show()
```

```python
# 7. Top Products
top_products = df.groupby('Product_ID')['Orders'].sum().sort_values(ascending=False).head(10)

plt.figure()
sns.barplot(x=top_products.values, y=top_products.index)
plt.title("Top 10 Products by Orders")
plt.show()
```
